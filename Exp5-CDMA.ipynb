```python
import numpy as np
from scipy.linalg import hadamard

# Generate 4 orthogonal Walsh codes
W = hadamard(4)

# Data bits of two users
bits1 = np.array([1, -1, 1, 1])
bits2 = np.array([-1, -1, 1, -1])

# Spreading function
def spread(bits, code):
    return np.concatenate([b * code for b in bits])

# Combine the spread signals
tx = spread(bits1, W[1]) + spread(bits2, W[2])

# Despreading function
def despread(rx, code):
    chips = rx.reshape(-1, len(code))
    return np.sign(chips @ code / len(code))

# Recover original data
print("User1 recovered:", despread(tx, W[1]), "sent:", bits1)
print("User2 recovered:", despread(tx, W[2]), "sent:", bits2)

# Check correlation using an incorrect code
print("Wrong code W3 correlation:",
      (tx.reshape(-1, 4) @ W[3]) / 4)
```

```python
import numpy as np
np.random.seed(1)
N_UE, N_RB, N_TTI = 4, 10, 1000
base = np.array([10, 7, 4, 1.5])

# Jain's fairness index
def jain(x):
    return x.sum()**2 / (len(x) * (x**2).sum())


# Run scheduler simulation
def run(policy):
    total = np.zeros(N_UE)
    avg = np.ones(N_UE)

    for t in range(N_TTI):

        # Generate channel fading
        rate = base * np.random.exponential(1, N_UE)

        got = np.zeros(N_UE)

        for rb in range(N_RB):

            if policy == "MaxCI":
                u = np.argmax(rate)

            elif policy == "RR":
                u = (rb + t) % N_UE

            else:
                u = np.argmax(rate / avg)

            got[u] += rate[u]

        # Update average throughput
        avg = 0.99 * avg + 0.01 * got

        total += got

    return total / N_TTI


# Compare all three scheduling policies
for p in ["MaxCI", "RR", "PF"]:

    x = run(p)

    print(
        f"{p:6} per-UE = {np.round(x, 1)} "
        f"Total = {x.sum():.1f} "
        f"Jain = {jain(x):.3f}"
    )
```

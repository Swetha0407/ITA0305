```python
import numpy as np
import matplotlib.pyplot as plt

# Mobile moves from BS1 (0 m) to BS2 (1000 m)
x = np.arange(0, 1001, 10)

# Calculate signal strength from both base stations
p1 = -30 - 35 * np.log10(np.maximum(x, 1))
p2 = -30 - 35 * np.log10(np.maximum(1000 - x, 1))

# Hysteresis margin and initial serving cell
H = 5
serving = 1

# Check handoff condition
for i in range(len(x)):
    if serving == 1 and p2[i] > p1[i] + H:
        serving = 2
        print(f"Handoff BS1 -> BS2 at {x[i]} m")
        break

# Plot signal strengths
plt.figure(figsize=(9, 5))
plt.plot(x, p1, label="BS1 Signal")
plt.plot(x, p2, label="BS2 Signal")

plt.xlabel("Distance (m)")
plt.ylabel("Signal Strength (dBm)")
plt.title("Handoff Simulation with Hysteresis")
plt.legend()
plt.grid(True)
plt.show()
```
